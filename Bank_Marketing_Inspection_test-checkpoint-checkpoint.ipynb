# Bank Marketing Data Analysis
This notebook presents an analysis of the `bankmarketing.csv` dataset, which includes data related to a bank's marketing campaigns. The main goal is to understand customer behavior and predict whether a client will subscribe to a term deposit.

```python

import pandas as pd

# Load the dataset
df = pd.read_csv('bankmarketing.csv')

# Display the first few rows
df.head()
```

```python

# Check for missing values and data types
df.isna().sum() # missing values check

 #chek data types
df.dtypes
```

```python

# Summary statistics
df.describe()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Plot the distribution of the target variable
sns.countplot(x='y',data=df)

plt.title("Targets Variable Distribution (y)")
plt.xlabel("Target Achieve (Yes/No)")
plt.ylabel("Count vaues")

plt.show()
```

```python

# Correlation matrix for numerical features

import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)
corr

plt.figure(figsize=(10,6))

sns.heatmap(corr,annot=True,cmap='viridis')

plt.title("Correlation Matrix",color='blue')
plt.show()
```
