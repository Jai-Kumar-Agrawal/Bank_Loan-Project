## Data Preprocessing
- Handle missing values
- Encode categorical variables
- Normalize numerical features

```python
#encoding categorical variables

category_cols=df.select_dtypes(include=object).columns
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

for col in category_cols:
    df[col]=le.fit_transform(df[col])
```

```python
# Drop unnecessary columns (e.g., Patient ID if present)
df.drop(['Patient_ID'],axis=1,inplace=True)
```

```python
#Normalized numeric feature
from sklearn.preprocessing import StandardScaler
cols=['Age','BMI','Time_to_Recurrence'] 
scaler = StandardScaler()
df[cols] = scaler.fit_transform(df[cols])
```

```python
# Adjust target column name as needed
```

```python
# Define features and target variable
x = df.drop('Survival_Status',axis=1)
y = df['Survival_Status']
```

```python
# Split into train and test sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

## Model Training
Train a **Logistic Regression** model to predict patient survival.

```python

# Train a Logistic Regression model

from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
lr.score(x_test,y_test)*100
```

```python
# Predictions
# Model evaluation
y_pred = lr.predict(x_test)
```

```python
# print("Accuracy:", accuracy_score(y_test, y_pred))

from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
# print("Classification Report:\n", classification_report(y_test, y_pred))
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
